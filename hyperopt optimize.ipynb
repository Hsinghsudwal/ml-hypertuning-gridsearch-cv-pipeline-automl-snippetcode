```python

Search Medium
Write

Harinder Singh Sudwal
Get unlimited access to all of Medium for less than $1/week.
Become a member


Optimizing Hyperparameters with Hyperopt: A Hands-On Tutorial
Learn How to Leverage Optuna to Optimize Hyperparameters
Okan Yenigün
Towards Dev
Okan Yenigün

·
Following

Published in
Towards Dev

·
7 min read
·
Jul 17
16







Photo by Drew Patrick Miller on Unsplash
Hyperopt is a Python library used for hyperparameter optimization, which is a crucial step in the process of machine learning model building. Hyperopt provides a framework for automating the search for optimal hyperparameters by employing different optimization algorithms.

Hyperopt utilizes a technique called Bayesian optimization, which intelligently explores the hyperparameter search space by leveraging past evaluations to guide future selections. It maintains a probabilistic model of the objective function based on the evaluated hyperparameter configurations and uses this information to decide which configuration to evaluate next. By iteratively evaluating and updating the model, Hyperopt narrows down the search space and converges towards the best hyperparameter settings.

Step-by-Step Guide to Bayesian Optimization: A Python-based Approach
Building the Foundation: Implementing Bayesian Optimization in Python
medium.com

The library offers several algorithms for hyperparameter optimization, including TPE (Tree-structured Parzen Estimator) and Random Search. TPE is the default algorithm used by Hyperopt and has shown good performance in many scenarios.

Exploring Tree Parzen Estimators for Hyperparameter Tuning
Brief Explanation and Implementation of TPEs
towardsdev.com

You can find the official documentation here.

pip install hyperopt

Basic Usage
We will optimize the (x-2)² function.

import numpy as np
from hyperopt import hp, fmin, tpe, Trials

def objective_function(x):
    return (x - 2) ** 2

space = hp.uniform('x', -10, 10)  # Define the search space for x

trials = Trials()  # Object to track the optimization process

best = fmin(
    fn=objective_function,
    space=space,
    algo=tpe.suggest,  # TPE algorithm for optimization
    max_evals=100,  # Maximum number of evaluations
    trials=trials
)

optimal_x = best['x']
optimal_value = objective_function(optimal_x)

print(f"Optimal x: {optimal_x}")
print(f"Optimal value: {optimal_value}")

"""
100%|███| 100/100 [00:00<00:00, 784.05trial/s, best loss: 0.0009664762470087422]
Optimal x: 1.9689117989100569
Optimal value: 0.0009664762470087422
"""
The objective function will simply return the value of (x-2)².

hp is a module that provides functions for defining the search space for hyperparameters.

hp.choice(label, options): Specifies a discrete choice among a set of options. The options argument should be a list or tuple of possible values. hp.choice("booster", ['gbtree', 'gblinear', 'dart'])
hp.uniform(label, low, high): Specifies a uniform distribution over the range from low to high. Hyperopt will sample values uniformly from this range. hp.uniform('learning_rate', 0.001, 0.2)
hp.quniform(label, low, high, q): Specifies a discrete uniform distribution over the range from low to high with a step size of q. Hyperopt will sample values that are integer multiples of q. hp.quniform('n_estimators', 100, 2000, 10)
hp.loguniform(label, low, high): Specifies a log-uniform distribution over the range from low to high. Hyperopt will sample values uniformly in the logarithmic space. hp.loguniform('min_child_weight', -5, 5)
hp.qloguniform(label, low, high, q): Specifies a discrete log-uniform distribution over the range from low to high with a step size of q. hp.qloguniform('subsample', np.log(0.5), np.log(1.0), 0.05)
hp.normal(label, mu, sigma): Specifies a normal distribution with mean mu and standard deviation sigma. Hyperopt will sample values from this normal distribution. hp.normal('colsample_byte', 0.5, 0.3)
trials = Trials()
Trials is a class that helps keep track of the results from different iterations (trials) of the hyperparameter optimization process. It provides a way to store and retrieve information about each trial, such as the hyperparameters used and the corresponding objective function values.

trials.trials - a list of dictionaries representing everything about the search
trials.results - a list of dictionaries returned by 'objective' during the search
trials.losses() - a list of losses (float for each 'ok' trial)
trials.statuses() - a list of status strings
The Trials object is typically passed as an argument to the optimization function, such as fmin(). During the optimization process, as Hyperopt evaluates different hyperparameter configurations, it updates the Trials object with the results of each trial.

best = fmin(
    fn=objective_function,
    space=space,
    algo=tpe.suggest,  # TPE algorithm for optimization
    max_evals=100,  # Maximum number of evaluations
    trials=trials
)
fmin performs the optimization of a given objective function over a search space defined by hyperparameters. It is one of the core functions provided by Hyperopt for hyperparameter tuning.

fn is the objective function.
The space variable represents the search space for the hyperparameters, defined using the hp functions from Hyperopt.
The algo argument specifies the optimization algorithm to be used, such as tpe.suggest, which stands for Tree-structured Parzen Estimator and is the default algorithm in Hyperopt. Other options include rand.suggest for random search or custom optimization algorithms.
The max_evals argument sets the maximum number of evaluations to be performed during the optimization process. Hyperopt will stop the optimization after reaching this number of evaluations.
The trials object is used to keep track of the optimization process. It collects the results from each trial, including the hyperparameters and the corresponding objective function values.
rstate (RandomState object): Allows you to provide a specific random state to control the randomness of the optimization process. By setting a fixed random state, you can reproduce the same results across multiple runs.
show_progressbar (bool): Determines whether to display a progress bar during the optimization process. When set to True, a progress bar will be shown to visualize the progress of the evaluations.
show_warnings (bool): Controls whether to display warnings during the optimization process. Setting this parameter to True enables the display of warning messages.
pass_expr_memo_ctrl (bool): Specifies whether to pass the ExprMemoControl object to the objective function. This object provides access to internal memoization control, which can be useful for advanced use cases.
return_argmin (bool): Determines whether the fmin function returns the best hyperparameters (False) or the index of the best hyperparameters in the trials (True). By default, it is set to False to return the best hyperparameters.
verbose (int): Controls the level of verbosity during the optimization process. Higher values (e.g., 1, 2) provide more verbose output for debugging purposes.
Machine Learning Hyperparameter Tuning
We will now perform an example of hyperparameter tuning in machine learning using the dataset containing laptop prices.

import pandas as pd
import xgboost as xgb
import numpy as np
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

df = pd.read_csv("laptopPrice.csv")
# there are some duplicates in the dataset
df = df.drop_duplicates()
df.reset_index(drop=True,inplace=True)

# some replacements

df['ram_gb'] = df['ram_gb'].str.replace(' GB', '').astype(int)
df['ssd'] = df['ssd'].str.replace(' GB', '').astype(int)
df['hdd'] = df['hdd'].str.replace(' GB', '').astype(int)
df['graphic_card_gb'] = df['graphic_card_gb'].str.replace(' GB', '').astype(int)
df['os_bit'] = df['os_bit'].str.replace('-bit', '').astype(int)
df["rating"] = df["rating"].replace({"1 star": "1 stars"})
df['rating'] = df['rating'].str.replace(' stars', '').astype(int)

mapping = {'ThinNlight': 1, 'Casual': 2, 'Gaming': 3}
df["weight"] = df["weight"].replace(mapping)

mapping = {'No warranty': 0, '1 year':1, '2 years':2, '3 years':3}
df["warranty"] = df["warranty"].replace(mapping)

mapping = {'No': 0, 'Yes': 1}
df["Touchscreen"] = df["Touchscreen"].replace(mapping)

mapping = {'No': 0, 'Yes': 1}
df["msoffice"] = df["msoffice"].replace(mapping)

df.head(10)

Dataset. Image by the author.
num_features = [feature for feature in df.columns if df[feature].dtype != 'object' and feature != "Price"]
cat_features = [feature for feature in df.columns if df[feature].dtype == 'object']

# feature transformation

cat_transformer = OneHotEncoder(handle_unknown='ignore')
num_transformer = StandardScaler()

preprocessor = ColumnTransformer(
    transformers=[
        ('cat', cat_transformer, cat_features),
        ('num', num_transformer, num_features)
    ])

X = preprocessor.fit_transform(df)
y = df["Price"].values.reshape(-1,1)

print(f"X: {X.shape}, y: {y.shape}")
# X: (802, 51), y: (802, 1)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, shuffle=True)
Once we have completed preprocessing and feature engineering, we have prepared our input data X and target variable y. Now, we will utilize an XGBoost regressor with its default configurations as our initial model.

model = xgb.XGBRegressor()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
rmse_train = np.sqrt(mean_squared_error(y_train, model.predict(X_train)))
rmse_test = np.sqrt(mean_squared_error(y_test, y_pred))
print(f"Train RMSE: {rmse_train}, Test RMSE: {rmse_test}")

# Train RMSE: 2712.4698978403653, Test RMSE: 18867.428818127133
Now, let’s employ Hyperopt to optimize and fine-tune specific hyperparameters.

from hyperopt import hp, fmin, tpe, Trials
from hyperopt.pyll.base import scope

# Define the objective function to optimize
def objective_function(params):
    # Create an XGBoost regressor with the hyperparameters from params
    model = XGBRegressor(**params)
    
    # Fit the model to the training data
    model.fit(X_train, y_train)
    
    # Make predictions on the test data
    y_pred = model.predict(X_test)
    
    # Calculate the mean squared error
    mse = mean_squared_error(y_test, y_pred)
    
    return mse

# Define the search space for hyperparameters
space = {
    'max_depth': scope.int(hp.quniform("n_estimators", 100, 10000, 100)),
    'max_depth': hp.choice('max_depth', [3, 5, 7, 9]),
    'learning_rate': hp.uniform('learning_rate', 0.01, 0.1),
    'subsample': hp.uniform('subsample', 0.5, 1.0),
    'colsample_bytree': hp.uniform('colsample_bytree', 0.5, 1.0)
}

# Initialize Trials to keep track of the optimization process
trials = Trials()

# Run the hyperparameter optimization
best = fmin(
    fn=objective_function,
    space=space,
    algo=tpe.suggest,
    max_evals=100,
    trials=trials
)

# Retrieve the best hyperparameters found
optimal_params = {k: v for k, v in best.items() if k in space}

# Train the final model with the optimal hyperparameters
final_model = XGBRegressor(**optimal_params)
final_model.fit(X_train, y_train)

print("Best Hyperparameters:", optimal_params)

"""
100%|████████| 100/100 [00:09<00:00, 10.87trial/s, best loss: 595134420.1607146]
Best Hyperparameters: {'colsample_bytree': 0.5989570380704331, 'learning_rate': 0.08046890770242751, 'max_depth': 3, 'subsample': 0.5817944941456368}
"""
When tuning hyperparameters using Hyperopt, certain hyperparameters, such as the number of estimators or the maximum depth of trees, may need to be integers. However, Hyperopt by default treats all hyperparameters as continuous variables. To specify that certain hyperparameters should be treated as integers, we can use the scope function from the Hyperopt library.

In conclusion, Hyperopt streamlines hyperparameter tuning with its intelligent exploration of search spaces and support for various algorithms. Its flexibility, parallel optimization, and seamless integration enhance model development, leading to better-performing machine learning models with reduced manual effort. Hyperopt is a valuable asset for building high-quality models efficiently.

Read More
Optimizing Hyperparameters with Optuna: A Hands-On Tutorial
Learn How to Leverage Optuna to Optimize Hyperparameters
pub.aimind.so

Optimizing Hyperparameters with Grid Search: A Hands-On Tutorial
Learn How to Leverage Grid Search to Optimize Hyperparameters
blog.devops.dev

Optimizing Hyperparameters with Random Search: A Hands-On Tutorial
Learn How to Leverage Random Search to Optimize Hyperparameters
awstip.com

Bayesian Modeling: From Prior Plausibilities to Posterior Distributions
Beyond Traditional Approaches: Exploring the Depths of Bayesian Inference in Complex Models
awstip.com

ANOVA Test in Python
Code ANOVA Statistical Test with Python
towardsdev.com

Sources
http://hyperopt.github.io/hyperopt/

https://www.youtube.com/watch?v=Mp1xnPfE4PY

Machine Learning
Python
Deep Learning
Hyperparameter Tuning
Optimization
16





Okan Yenigün
Towards Dev
Written by Okan Yenigün
988 Followers
·
Writer for 
Towards Dev

Top Writer for Django | Design Patterns | Image Processing | Regression | Time Series Forecasting. Writes about Python | Scala | Go | ML&DL |

Following

More from Okan Yenigün and Towards Dev
A Catalog For Design Patterns in Python
Okan Yenigün
Okan Yenigün

in

Towards Dev

A Catalog For Design Patterns in Python
Exploring Design Patterns: Concepts, Analogies, Advantages, Disadvantages, and Real-Life Use Cases.
19 min read
·
Mar 16
135



Common Mistakes to Avoid in Jetpack Compose
Android-World
Android-World

in

Towards Dev

Common Mistakes to Avoid in Jetpack Compose
Photo by Santa Barbara on Unsplash

·
3 min read
·
Apr 20
305



Data Engineering Project — Real Time Weather Prediction
Bar Dadon
Bar Dadon

in

Towards Dev

Data Engineering Project — Real Time Weather Prediction
Introduction
14 min read
·
Jul 21
85



Design Patterns in Python: State Pattern
Okan Yenigün
Okan Yenigün

in

Dev Genius

Design Patterns in Python: State Pattern
The State Design Pattern Explained and Implemented in Python
3 min read
·
Mar 14
51



See all from Okan Yenigün
See all from Towards Dev
Recommended from Medium
Churn prediction using Ensemble Techniques-II
Vignesh Gopalakrishnan
Vignesh Gopalakrishnan

Churn prediction using Ensemble Techniques-II
Overview
12 min read
·
Jul 23
1



HTML REPORT
Prathamesh Gadekar
Prathamesh Gadekar

in

Level Up Coding

Python Libraries for Lazy Data Scientists
Do you feel lethargic today? Use these five libraries to boost your productivity.
7 min read
·
Apr 7
477

2



Lists



Predictive Modeling w/ Python
20 stories
·
282 saves
Principal Component Analysis for ML
Time Series Analysis
deep learning cheatsheet for beginner
Practical Guides to Machine Learning
10 stories
·
294 saves



Natural Language Processing
522 stories
·
141 saves



Coding & Development
11 stories
·
110 saves
Managing Long-Running Processes with FastAPI in Python
Kodingsenpai
Kodingsenpai

in

Stackademic

Managing Long-Running Processes with FastAPI in Python
Understanding Long-Running Processes
2 min read
·
Aug 12
21



Using optuna with sklearn the right way — Part 1
Walter Sperat
Walter Sperat

Using optuna with sklearn the right way — Part 1
Using optuna with sklearn correctly can get ugly fast, let's try and make it simpler
11 min read
·
Jul 11
68

2



Multiclass classification using CatBoost and SHAP in Python
lochie links
lochie links

Multiclass classification using CatBoost and SHAP in Python
What is Catboost and SHAP
9 min read
·
Jul 12
21

1



Conformal Prediction: An easy way to estimate prediction intervals
María Jesús Ugarte
María Jesús Ugarte

in

bain-inside-advanced-analytics

Conformal Prediction: An easy way to estimate prediction intervals
Uncertainty for single predictions is an important issue in machine learning nowadays, specially when a wrong prediction could bring bad…
5 min read
·
Jul 11
16



See more recommendations
```
